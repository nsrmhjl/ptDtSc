## Dictionaries

```python
from collections import defaultdict

s = "We tried list and we tried dicts also we tried Zen"

s = s.split()

counts = defaultdict(int)

for word in s:
    counts[word] += 1

for word, count in counts.items():
    print(word, " ", count)
```

## Complementing a Strand of DNA

```python
s= "AAAACCCGGT"
r = s[::-1]

comp = {"A": "T", "T": "A", "C": "G", "G": "C"}

rc = ""
for letter in r:
    rc += comp[letter]

print(rc)
```

## Computing GC Content

```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

from collections import defaultdict

gc = SeqIO.parse("gc.txt", "fasta")

gcFr = {}

for seq_record in gc:
    gcFr[gc_fraction(seq_record.seq)] = seq_record.id

print(gcFr[max(gcFr.keys())])
print(round(max(gcFr.keys()) * 100, 7))

```
